```python
!pip install pennylane pennylane-ionq -q

import pennylane as qml
from pennylane import numpy as np
from google.colab import userdata

# ==========================================
# 1. ENVIRONMENT & DEVICE CONFIGURATION
# ==========================================
ionq_token = userdata.get("IONQ_AF")

# Expanded register allocation for 3p orbital distributions
num_wires = 7
dev_sim = qml.device('ionq.simulator', wires=num_wires, api_key=ionq_token)

# ==========================================
# 2. EXPANDED MASTER ATOMIC CIRCUIT ARCHITECTURE
# ==========================================
@qml.qnode(dev_sim, shots=1024)
def master_atomic_circuit(element, field_params):
    """
    Unified Master Quantum Circuit handling multi-electron configurations
    dynamically across the expanded 3p periodic table blocks.
    """
    el = element.lower().strip()

    # ------------------------------------------
    # LITHIUM (Z=3) | 3 Wires
    # ------------------------------------------
    if el == "lithium":
        qml.Hadamard(wires=0)
        qml.CNOT(wires=[0, 1])
        qml.RX(field_params[0], wires=2)
        qml.RZ(field_params[1], wires=2)

    # ------------------------------------------
    # BERYLLIUM (Z=4) | 4 Wires
    # ------------------------------------------
    elif el == "beryllium":
        qml.Hadamard(wires=0)
        qml.CNOT(wires=[0, 1])
        qml.RX(field_params[0], wires=2)
        qml.RX(field_params[1], wires=3)
        qml.CNOT(wires=[2, 3])
        qml.RZ(field_params[2], wires=3)

    # ------------------------------------------
    # MAGNESIUM (Z=12) | 3 Wires
    # ------------------------------------------
    elif el == "magnesium":
        qml.RX(field_params[0], wires=0)
        qml.RY(field_params[1], wires=1)
        qml.CNOT(wires=[0, 1])
        qml.RZ(field_params[2], wires=2)

    # ------------------------------------------
    # SILICON CORE (Z=14) | 6 Wires
    # ------------------------------------------
    elif el == "silicon":
        # Neon Core Simulation (Qubits 0-2)
        qml.Hadamard(wires=0)
        qml.CNOT(wires=[0, 1])
        qml.CNOT(wires=[1, 2])
        # Valence 3s Subshell
        qml.RX(field_params[0], wires=3)
        qml.RX(field_params[1], wires=4)
        qml.CNOT(wires=[3, 4])
        # Valence 3p Subshell Dynamic Coupling
        qml.RY(field_params[2], wires=5)
        qml.CNOT(wires=[4, 5])
        qml.RZ(field_params[3], wires=5)

    # ------------------------------------------
    # PHOSPHORUS (Z=15) | 6 Wires (Unpaired 3p Electrons)
    # ------------------------------------------
    elif el == "phosphorus":
        # Closed Neon Core Vector
        qml.Hadamard(wires=0)
        qml.CNOT(wires=[0, 1])

        # 3s2 Closed Subshell State
        qml.RX(field_params[0], wires=2)
        qml.RY(field_params[1], wires=3)
        qml.CNOT(wires=[2, 3])

        # 3p3 Half-Filled Shell (Hund's Rule parallel states: Wires 4, 5, 6)
        qml.RX(field_params[2], wires=4)
        qml.RX(field_params[3], wires=5)
        qml.RX(field_params[4], wires=6)

        # Global magnetic resonance coupling across the 3p orbitals
        qml.CNOT(wires=[4, 5])
        qml.CNOT(wires=[5, 6])
        qml.RZ(field_params[5], wires=6)

    # ------------------------------------------
    # SULFUR (Z=16) | 7 Wires (Paired 3p Electron)
    # ------------------------------------------
    elif el == "sulfur":
        # Closed Neon Core Vector
        qml.Hadamard(wires=0)
        qml.CNOT(wires=[0, 1])

        # 3s2 Closed Subshell State
        qml.RX(field_params[0], wires=2)
        qml.RY(field_params[1], wires=3)
        qml.CNOT(wires=[2, 3])

        # 3p4 Subshell Configuration
        # Wires 4 & 5 represent the newly paired 3p_x orbital
        qml.RX(field_params[2], wires=4)
        qml.RY(field_params[3], wires=5)
        qml.CNOT(wires=[4, 5])

        # Wires 6 & 7 represent the remaining unpaired 3p_y and 3p_z states
        qml.RX(field_params[4], wires=6)

        # Cross-orbital dynamic coupling network
        qml.CNOT(wires=[5, 6])
        qml.RZ(field_params[5], wires=6)

    else:
        raise ValueError(f"Element '{element}' layout not yet defined in the circuit profile.")

    return [qml.expval(qml.PauliZ(i)) for i in range(num_wires)]

# ==========================================
# 3. VERIFICATION RUNS
# ==========================================
phosphorus_dials = np.array([0.21, 0.65, 1.43, 0.88, 0.12, 0.95], requires_grad=True)
sulfur_dials = np.array([0.33, 0.72, 1.15, 0.44, 0.89, 0.61], requires_grad=True)

print("--- Running Phosphorus Simulation ---")
p_results = master_atomic_circuit("phosphorus", phosphorus_dials)
print(f"Phosphorus Expectations: {p_results}\n")

print("--- Running Sulfur Simulation ---")
s_results = master_atomic_circuit("sulfur", sulfur_dials)
print(f"Sulfur Expectations: {s_results}")
```

```python
# Phase 3: Unified-Field-Control Core Quantum Circuit Implementation
# Designed for execution via Google Colab / Local Terminal Link

!pip install qiskit qiskit-aer numpy -q

import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

def build_chsh_circuit(theta_A, theta_B):
    """
    Constructs a 2-qubit circuit to test the CHSH inequality.
    Applies a rotation to map specific spatial measurement angles.
    """
    # 2 Quantum registers, 2 Classical registers
    qc = QuantumCircuit(2, 2)

    # Step 1: Create maximally entangled Bell State |\Phi^+>
    qc.h(0)
    qc.cx(0, 1)

    # Step 2: Apply basis rotations for measurement
    # Qubit 0 rotation for Observer A
    qc.ry(2 * theta_A, 0)
    # Qubit 1 rotation for Observer B
    qc.ry(2 * theta_B, 1)

    # Step 3: Measure both qubits
    qc.measure([0, 1], [0, 1])

    return qc

# Define the optimal CHSH angles (in radians) for maximum violation
angles = {
    'A_B':   (0,             np.pi / 8),     # 0 and 22.5 deg
    'A_Bprime': (0,          -np.pi / 8),    # 0 and -22.5 deg
    'Aprime_B': (np.pi / 4,  np.pi / 8),     # 45 and 22.5 deg
    'Aprime_Bprime': (np.pi / 4, -np.pi / 8) # 45 and -22.5 deg
}

# Setup local simulator for logic validation before routing to IonQ QPU
simulator = AerSimulator()

print("Initializing Unified-Field-Control Simulation Verification...")
for setup, (theta_a, theta_b) in angles.items():
    circuit = build_chsh_circuit(theta_a, theta_b)
    compiled_circuit = transpile(circuit, simulator)

    # Execute batch job
    job = simulator.run(compiled_circuit, shots=2048)
    result = job.result()
    counts = result.get_counts()

    print(f"Setup {setup} (Angles A:{np.degrees(theta_a)}°, B:{np.degrees(theta_b)}°) -> Counts: {counts}")
```

```python
def calculate_chsh_s_value(counts_ab, counts_ab_prime, counts_a_prime_b, counts_a_prime_b_prime, shots):
    """
    Calculates the CHSH 'S' value from the measurement counts for the four different settings.
    E(X, Y) = (P(00) + P(11) - P(01) - P(10)) / total_shots
    S = E(A, B) + E(A, B') + E(A', B) - E(A', B')
    """
    def get_expectation_value(counts, total_shots):
        p00 = counts.get('00', 0)
        p01 = counts.get('01', 0)
        p10 = counts.get('10', 0)
        p11 = counts.get('11', 0)
        return (p00 + p11 - p01 - p10) / total_shots

    e_ab = get_expectation_value(counts_ab, shots)
    e_ab_prime = get_expectation_value(counts_ab_prime, shots)
    e_a_prime_b = get_expectation_value(counts_a_prime_b, shots)
    e_a_prime_b_prime = get_expectation_value(counts_a_prime_b_prime, shots)

    # Corrected formula for CHSH S value based on the standard Bell operator
    s_value = e_ab + e_ab_prime + e_a_prime_b - e_a_prime_b_prime
    return s_value

# Store counts for each setup
all_counts = {}

print("Recalculating CHSH simulation and storing counts...")
for setup, (theta_a, theta_b) in angles.items():
    circuit = build_chsh_circuit(theta_a, theta_b)
    compiled_circuit = transpile(circuit, simulator)

    # Execute batch job
    job = simulator.run(compiled_circuit, shots=2048)
    result = job.result()
    counts = result.get_counts()
    all_counts[setup] = counts
    print(f"Setup {setup} (Angles A:{np.degrees(theta_a)}°, B:{np.degrees(theta_b)}°) -> Counts: {counts}")


# Assuming 2048 shots were used for each experiment as defined in the original code
shots = 2048

# Get counts for each of the four settings
counts_ab = all_counts['A_B']
counts_ab_prime = all_counts['A_Bprime']
counts_a_prime_b = all_counts['Aprime_B']
counts_a_prime_b_prime = all_counts['Aprime_Bprime']

# Calculate the CHSH S value
chsh_s_value = calculate_chsh_s_value(counts_ab, counts_ab_prime, counts_a_prime_b, counts_a_prime_b_prime, shots)

print(f"\nCalculated CHSH S value: {chsh_s_value}")

# According to Bell's theorem, for classical systems, S <= 2.
# For quantum systems, S can violate this bound, up to 2*sqrt(2) (approx 2.828).
```
